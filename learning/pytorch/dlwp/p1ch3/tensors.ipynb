### Named tensors
Ejemplo imágenes

```python
img_t = torch.randn(3, 5, 5) # shape [channels, rows, columns] -> RGB image with 5x5 pixels
weights = torch.tensor([0.2126, 0.7152, 0.0722]) # weights for RGB to grayscale conversion
batch_t = torch.randn(2, 3, 5, 5) # shape [batch, channels, rows, columns] -> batch of 2 RGB images with 5x5 pixels
img_t.shape, batch_t.shape
```

```python
# from RGB to grayscale using weighted sum
# RGB is always the last dimension (-3)
img_gray_naive = img_t.mean(-3)
batch_gray_naive = batch_t.mean(-3)
img_gray_naive.shape, batch_gray_naive.shape
```

```python
img_t, batch_t
```

```python
img_gray_naive, batch_gray_naive
```

```python
unsqueezed_weights = weights.unsqueeze(-1).unsqueeze(-1) # add two dimensions at the end
weights, unsqueezed_weights, weights.shape, unsqueezed_weights.shape
```

```python
img_weights = img_t * unsqueezed_weights
batch_weights = batch_t * unsqueezed_weights
img_gray_weighted = img_weights.sum(-3)
batch_gray_weighted = batch_weights.sum(-3)
img_gray_weighted.shape, batch_gray_weighted.shape
```

_einsum_

```python
img_gray_weighted_fancy = torch.einsum("...chw,c->...hw", img_t, weights)
batch_gray_weighted_fancy = torch.einsum("...chw,c->...hw", batch_t, weights)
batch_gray_weighted_fancy.shape
```

Nombrando dimensiones

```python
weights_named = torch.tensor([0.2126, 0.7152, 0.0722], names=["channels"])
weights_named.shape, weights_named.names
```

```python
img_named = img_t.refine_names(..., "channels", "rows", "columns")
batch_named = batch_t.refine_names(..., "channels", "rows", "columns")
print(f"img named: {img_named.shape}, {img_named.names}")
print(f"batch named: {batch_named.shape}, {batch_named.names}")
```

_align_as_

```python
weights_aligned = weights_named.align_as(img_named)
weights_aligned.shape, weights_aligned.names
```

```python
gray_named = (img_named * weights_aligned).sum("channels")
gray_named.shape, gray_named.names
```

```python
# gray_named = (img_named[..., :3] * weights_named).sum("channels")
```

```python
gray_plain = gray_named.rename(None)
gray_plain.shape, gray_plain.names
```

Hacer cambios sobre un "subtensor" proveniente de otro implica que los cambios también se hagan sobre el tensor original

```python
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
second_point = points[1]
second_point[0] = 10.0
points
```

Cuando no se desee que esto suceda se deben crear nuevos tensores

```python
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
second_point = points[1].clone()
second_point[0] = 10.0
points
```

Ambos tensores comparten el mismo _storage_ (?)

```python
id(points.storage()) == id(points_t.storage())
```

```python
points.stride(), points_t.stride()
```

```python
points = torch.tensor([[4.0, 1.0], [5.0, 3.0], [2.0, 1.0]])
points_t = points.t()

print("id(storage):", id(points.storage()), id(points_t.storage()))
print("data_ptr:", points.data_ptr(), points_t.data_ptr())
print("comparten memoria (data_ptr equal)?", points.data_ptr() == points_t.data_ptr())
print("storage_offset:", points.storage_offset(), points_t.storage_offset())
print("stride:", points.stride(), points_t.stride())
```

```python
# measure times cpu vs gpu
import time
def measure_matrix_mult(size, device):
    a = torch.randn(size, device=device)
    b = torch.randn(size, device=device)
    start_time = time.time()
    a @ b
    end_time = time.time()
    return end_time - start_time

size = (10000, 10000)
cpu_time = measure_matrix_mult(size, device='cpu')
print(f"CPU time: {cpu_time:.6f} seconds")
gpu_time = measure_matrix_mult(size, device='cuda')
print(f"GPU time: {gpu_time:.6f} seconds")
```
