```python
import imageio.v3 as imageio # using imageio v3 to avoid deprecation warnings

img_dog_png = imageio.imread("img/dog.png")
img_dog_jpg = imageio.imread("img/dog.jpg")

print(f"PNG image shape: {img_dog_png.shape}") # png format has 4 channels (RGBA) - RGB + Alpha channel
print(f"JPG image shape: {img_dog_jpg.shape}") # jpg format has 3 channels (RGB)
```

Las imágenes leídas desde _imageio_ se representan como un _array_ de _numpy_ de forma (W, H, C). Para que puedan utilizarse como tensores de _pytorch_ es necesario modificar la forma para que sea (C, H, W)

```python
import torch

img1 = torch.from_numpy(img_dog_jpg)
img1_permuted = img1.permute(2, 0, 1)  # change from (W, H, C) to (C, H, W)
print(f"Permuted JPG image shape: {img1_permuted.shape}")
```
