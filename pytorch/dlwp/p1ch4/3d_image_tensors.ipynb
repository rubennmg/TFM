```python
import imageio

path = "img/3d_image_example"
vol_arr = imageio.volread(path, format="DICOM")
print(f"Volume shape: {vol_arr.shape}, dtype: {vol_arr.dtype}")
```

```python
import torch

vol = torch.from_numpy(vol_arr)
vol = torch.unsqueeze(vol, dim=0)  # Add batch dimension
print(f"Volume tensor shape: {vol.shape}, dtype: {vol.dtype}")
```
